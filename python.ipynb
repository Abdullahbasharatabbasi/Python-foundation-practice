```python
# Variables and basic data types
name = "Ali"  # str
age = 21      # int
height = 5.9  # float
is_student = True  # bool

print(name, age, height, is_student)
print(type(name), type(age), type(height), type(is_student))
```

Operators

```python
a, b = 10, 3
print("Add:", a + b)
print("Sub:", a - b)
print("Mul:", a * b)
print("Div:", a / b)
print("Floor Div:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
print("Comparison:", a > 2, a == b)
print("Logical:", (a > 5) and (b < 5))
```
